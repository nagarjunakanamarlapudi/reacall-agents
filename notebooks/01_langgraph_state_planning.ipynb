EDUCATIONAL — SELF-CONTAINED

This lesson makes an explicit LangGraph `StateGraph` visible. A small recall question moves through typed state, planning, a node, and a conditional edge. Every value is local teaching data; the graph does not contact a model or a service.

```python
from typing import Literal, TypedDict

from langgraph.graph import END, START, StateGraph


class RecallState(TypedDict):
    question: str
    plan: list[str]
    evidence: list[str]
    route: str


def plan_node(state: RecallState) -> dict:
    plan = ["extract predicate", "match products and lots", "reconcile units"]
    print("plan_node ->", plan)
    return {"plan": plan, "route": "investigate"}


def investigate_node(state: RecallState) -> dict:
    evidence = ["notice says plant P-1950", "lot L-157 is an exact date match"]
    print("investigate_node ->", evidence)
    return {"evidence": evidence}


def review_node(state: RecallState) -> dict:
    print("review_node -> conditional review")
    return {"route": "review"}


def choose_next(state: RecallState) -> Literal["investigate", "review"]:
    return "investigate" if state["route"] == "investigate" else "review"


builder = StateGraph(RecallState)
builder.add_node("plan", plan_node)
builder.add_node("investigate", investigate_node)
builder.add_node("review", review_node)
builder.add_edge(START, "plan")
builder.add_conditional_edges("plan", choose_next, {"investigate": "investigate", "review": "review"})
builder.add_edge("investigate", "review")
builder.add_edge("review", END)
graph = builder.compile()

result = graph.invoke({"question": "Which lots need a hold?", "plan": [], "evidence": [], "route": ""})
print("final state ->", result)
assert result["plan"][0] == "extract predicate"
assert result["evidence"] == ["notice says plant P-1950", "lot L-157 is an exact date match"]
print("ASSERTION PASSED: state flowed through plan, conditional route, and review")
```

```python
print("EDUCATIONAL — SELF-CONTAINED")
print("State is data; nodes return updates; edges select the next node.")
assert "plan" in result and "evidence" in result
print("ASSERTION PASSED: EDUCATIONAL — SELF-CONTAINED")
```
