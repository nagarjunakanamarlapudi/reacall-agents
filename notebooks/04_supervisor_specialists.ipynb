EDUCATIONAL — SELF-CONTAINED

A supervisor owns the shared case question and delegates bounded tasks to specialists. `write_todos` represents the planning contract used by Deep Agents concepts: explicit work, completion criteria, and delegation. Specialists receive only their task context (context quarantine).

```python
def write_todos(question: str) -> list[dict]:
    return [
        {"task": "regulatory intake", "done_when": "predicate has product, plant, and dates"},
        {"task": "product and lot matching", "done_when": "each candidate is classified"},
        {"task": "trace and reconcile", "done_when": "received equation balances or gap is recorded"},
    ]


def regulatory_specialist(task: dict) -> dict:
    assert set(task) == {"recall_number"}
    return {"specialist": "regulatory", "predicate": {"plant": "P-1950", "julian": [157, 184]}}


def matching_specialist(task: dict) -> dict:
    assert set(task) == {"product", "plant"}
    return {"specialist": "matching", "classification": "ambiguous", "lot": task["product"]}


def trace_specialist(task: dict) -> dict:
    assert set(task) == {"lot", "events"}
    received = sum(event["quantity"] for event in task["events"] if event["kind"] == "receive")
    return {"specialist": "traceability", "received": received, "gap": received != 10}


todos = write_todos("Which recalled units remain unresolved?")
print("write_todos ->", todos)
assignments = {
    "regulatory": regulatory_specialist({"recall_number": "H-1230-2026"}),
    "matching": matching_specialist({"product": "SKU-1", "plant": "P-1950"}),
    "traceability": trace_specialist({"lot": "L-157", "events": [{"kind": "receive", "quantity": 10}]}),
}
print("supervisor assignments ->", assignments)
assert len(todos) == 3 and todos[0]["done_when"]
assert assignments["matching"]["classification"] == "ambiguous"
assert assignments["traceability"]["received"] == 10
print("ASSERTION PASSED: supervisor delegated bounded, quarantined specialist contexts")
```

```python
print("EDUCATIONAL — SELF-CONTAINED")
print("Deep Agents adds planning, delegation, and workspace context; the outer graph still owns safety decisions.")
assert len(assignments) == 3 and len(todos) == 3
print("ASSERTION PASSED: EDUCATIONAL — SELF-CONTAINED")
```
