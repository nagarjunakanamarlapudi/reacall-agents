This compact end-to-end investigation embeds an authoritative-looking recall predicate and a clearly synthetic retailer twin. It matches products and lots, traces units, performs the reconciliation equation, and evaluates safety-critical outcomes. It is an educational simulation, not a live public-health decision.

```python
recall = {"number": "H-1230-2026", "plant": "P-1950", "julian_range": range(157, 185), "hazard": "Salmonella", "origin": "PUBLIC_SNAPSHOT"}
products = [
    {"sku": "SKU-1", "plant": "P-1950", "julian": 157, "lot": "L-157", "origin": "SYNTHETIC_RETAILER_DIGITAL_TWIN"},
    {"sku": "SKU-2", "plant": "P-9999", "julian": 157, "lot": "L-157X", "origin": "SYNTHETIC_RETAILER_DIGITAL_TWIN"},
]
events = [
    {"lot": "L-157", "kind": "receive", "quantity": 10, "facility": "DC-West"},
    {"lot": "L-157", "kind": "ship", "quantity": 6, "facility": "Store-7"},
    {"lot": "L-157", "kind": "quarantine", "quantity": 2, "facility": "DC-West"},
    {"lot": "L-157", "kind": "sold", "quantity": 2, "facility": "Store-7"},
]

matched = [product for product in products if product["plant"] == recall["plant"] and product["julian"] in recall["julian_range"]]
print("predicate ->", {key: value for key, value in recall.items() if key != "julian_range"})
print("matched products ->", matched)
received = sum(event["quantity"] for event in events if event["kind"] == "receive")
on_hand = 6
quarantined = sum(event["quantity"] for event in events if event["kind"] == "quarantine")
sold = sum(event["quantity"] for event in events if event["kind"] == "sold")
returned = disposed = 0
unaccounted = received - (on_hand + quarantined + sold + returned + disposed)
reconciliation = {"received": received, "on_hand": on_hand, "quarantined": quarantined, "sold": sold, "returned": returned, "disposed": disposed, "unaccounted": unaccounted}
print("reconciliation ->", reconciliation)
facilities = sorted({event["facility"] for event in events})
metrics = {
    "predicate_match": matched[0]["sku"] == "SKU-1",
    "facility_coverage": set(facilities) == {"DC-West", "Store-7"},
    "reconciliation_gap_detected": unaccounted == 0,
    "closure_allowed": unaccounted == 0,
}
metrics.update({"tool_calls": 4, "steps": 4, "elapsed_ms": 12, "reliability": "deterministic"})
print("evaluation ->", metrics)
assert len(matched) == 1 and matched[0]["lot"] == "L-157"
assert reconciliation["received"] == 10 and reconciliation["unaccounted"] == 0
assert metrics["predicate_match"] and metrics["facility_coverage"]
assert metrics["closure_allowed"]
print("ASSERTION PASSED: investigation and safety evaluation reached a balanced, closeable case")
```
